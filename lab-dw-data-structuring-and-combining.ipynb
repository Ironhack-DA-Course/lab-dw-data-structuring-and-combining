```python
def clean_State(x):
    '''Replace abbreviation with full name'''
    match x:
        case "Cali":
            return "California"
        case "WA":
            return "Washington"
        case "AZ":
            return "Arizona"
        case _:
            return x

def clean_Gender(x):
    match x:
        case "Femal":
            return "F"
        case "Male":
            return "M"
        case "Femal":
            return "F"
        case _:
            return x
def clean_CTV(x):
    # if x != None:
        x = str(x)
        if x.endswith("%"):
            # return float(x_str[0:len(x)-1])
            # return float(x_str.str.replace(r"-\%", "", regex=True))
            return float(x.replace("%", ""))

        else:
            return x
    # else:
    #     return x
df2 = df_all.copy()        
df2["state"] = df2["state"].apply(clean_State)
df2["gender"] = df2["gender"].apply(clean_Gender)
df2["education"] = df2["education"].apply(lambda x: "Bachelor" if x == "Bachelors" else x)
df2["vehicle_class"] = df2["vehicle_class"].apply(lambda x: "Luxury" if x in ["Sports Car", "Luxury SUV", "Luxury Car"] else x) 
df2["customer_lifetime_value"] = df2["customer_lifetime_value"].apply(clean_CTV)
```
